## Python数据分析处理工具库Pandas的使用操作

参考菜鸟教程：https://www.runoob.com/pandas/pandas-tutorial.html

（1）学习pandas中的DataFrame和Series两种数据结构的使用。
Dataframe：https://www.runoob.com/pandas/pandas-dataframe.html

导入库，创建DataFrame并输出，代码如下

```python
import pandas as pd
data = {
    '产品': ['手机', '笔记本', '平板', '耳机', '智能手表', 
           '相机', '电视', '音响', '打印机', '路由器'],
    '销售量': [150, 80, 90, 200, 120, 
              60, 50, 130, 70, 40],
    '价格(元)': [3000, 6000, 2000, 800, 1500, 
                5000, 4000, 1500, 1000, 600],
    '销售日期': pd.date_range(start='2023-01-01', periods=10, freq='D')
}
df = pd.DataFrame(data)
print(df)
```

在Jupyter notebook中实现：

（1）筛选出销售量大于 100 的产品

```python
df1=df[df['销售量']>100]
df1
```

（2）按销售量从大到小排序并输出

```python
df2=df.sort_values(by='销售量',ascending=False)
df2
```

（3）计算所有产品的平均销售量

```python
print('所有产品的平均销售量:',df['销售量'].mean())
```

（4）计算每种产品的总销售额，并添加总销售额为新列

```python
df['总销售额']=df['销售量']*df['价格(元)']
df
```

## Python可视化工具Matplotlib的简单使用操作

从菜鸟教程：https://www.runoob.com/matplotlib/matplotlib-tutorial.html

学习掌握常用的绘图方法，绘图标记、线、轴标签和标题、网格线

（1）示例1 绘制余弦函数$cos(x)$

```python
import matplotlib.pyplot as plt
import numpy as np
plt.rcParams['font.sans-serif'] = ['SimHei'] # 显示中文
plt.rcParams['axes.unicode_minus'] = False # 显示负号
x = np.linspace(0, 2 * np.pi, 100)
y = np.cos(x)

plt.plot(x, y, color='orange')
plt.title('余弦函数')
plt.xlabel('x')
plt.ylabel('cos(x)')
plt.grid()
# jupyter notebook中不需要加plt.show()
```

（2）示例2 绘制条形图

```python
categories = ['A', 'B', 'C', 'D']
values = [4, 7, 1, 8]

plt.bar(categories, values, color='skyblue')
plt.title('条形图')
plt.xlabel('类别')
plt.ylabel('值')
```

在Jupyter notebook中实现：

（1）绘制复合函数$y=sin(x)*cos(x)$

```python
x = np.linspace(0, 2 * np.pi, 100)
y = np.sin(x)*np.cos(x)

plt.plot(x, y, color='orange')
plt.title('余弦函数')
plt.xlabel('x')
plt.ylabel('$sin(x)*cos(x)$') # 使用LaTeX语法
plt.grid()
```
