# 实验5-1 感知机

假设银行需要根据顾客的年龄、薪资、当前债务来决定是否给予信用卡，已知5个人获得信用卡时的相关信息如下：

| 指标  | 客户1 | 客户2 | 客户3 | 客户4 | 客户5 | 自己 |
| ----- | ----- | ----- | ----- | ----- | ----- | ----- |
|年龄| 16 | 24 | 23 | 28 | 40 | 50 |
|薪资（月）| 0 | 5000 | 6000 | 5000 | 15000 | 12000 |
|债务（万）| 0 | 0 | 10 | 50 | 200 | 30| 
|是否获批| 否 | 是 | 是 | 否 | 是 | ？ |

```python
import numpy as np
from tqdm import tqdm
```

1. 输入变量`X=[年龄，薪资，债务]`

```python
# 数据输入：年龄, 薪资（月）, 债务（万） 
X = np.array([[16, 0, 0],
              [24, 5000, 0],
              [23, 6000, 10],
              [28, 5000, 50],
              [40, 15000, 200]])
# 目标输出（0: 否, 1: 是） 
y = np.array([0, 1, 1, 0, 1])
```

2. 初始化权重`W=[w1,w2,w3]`和偏置值为`b`

```python
W = np.array([0.1, 0.1, 0.1])
b = 0.1
```

3. 激活函数选择$sigmoid$，输出$y=sigmod(w_1*x_1+w_2*x_2+w_3*x_3+b)$

```python
def sigmoid(x):
    return 1 / (1 + np.exp(-x))
```

4. 手工尝试不同的参数，比较误差值

```python
W= np.array([0.1, 0.1, 0.1])
b = 0.1
print('y_pred:', sigmoid(np.dot(X, W) + b))
W= np.array([1, 1, 1])
b = 0.1
print('y_pred:', sigmoid(np.dot(X, W) + b))
W= np.array([0.1, 0.6, 0.3])
b = 0
print('y_pred:', sigmoid(np.dot(X, W) + b))
```

5. 实现梯度下降求解，更新权重和偏置，并记录最终的权重和偏置


```python
def gradient_descent(X, y, weights, bias, lr=0.01, epochs=10):
    for _ in tqdm(range(epochs)):
        # 计算y的预测值
        y_pred = sigmoid(np.dot(X, weights) + bias)
        # 计算损失
        loss = -np.mean(y * np.log(y_pred) + (1 - y) * np.log(1 - y_pred))
        # 计算损失对权重的偏导数
        dW = np.dot(X.T, y_pred - y) / len(y)
        # 计算损失对偏置的偏导数
        db = np.mean(y_pred - y)
        # 更新权重
        weights -= lr * dW
        # 更新偏置
        bias -= lr * db
    return weights, bias


W, b = gradient_descent(X, y, W, b)
print('权重:', W)
print('偏置:', b)
```

检验预测结果

```python
sigmoid(np.dot(X, W) + b)
```

6. 判断自己是否获批。当输出`y<0.5`，未获批，否则获批

```python
X = np.array([[50, 12000, 30]])
y_pred = sigmoid(np.dot(X, W) + b)
print('预测值:', y_pred[0])
if y_pred[0] >= 0.5:
    print('获批')
else:
    print('未获批')
```
